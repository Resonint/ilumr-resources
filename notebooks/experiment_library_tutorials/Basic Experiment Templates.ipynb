For more information on how to override pulse sequence parameters, create user input widgets, and customise the GUI, please also see our Experiment Class Tutorial.

This tutorial covers:

## State Cache

When you create an MRI experiment using the matipo experiment library (`BaseExperiment`), plot data and parameter values from the experiment class will be saved into a state cache and reloaded into the experiment when the Jupyter notebook is opened. When the experiment object is created, you should assign a unique state ID.

## Pulse Collect (FID) with Signal and Spectrum Plots

The following code is for a very simple implementation of a pulse collect experiment with signal and spectrum plots. 

In this experiment, the Free Induction Decay pulse sequence (`FID.py`) is loaded from *ilumr's* global sequence directory.

The experiment uses the default pulse sequence parameters in the `FID.py` pulse sequence file.

The data is displayed using `SignalPlot` and `SpectrumPlot` from the *matipo* experiment library.

```python
from matipo import Sequence, SEQUENCE_DIR
from matipo.experiment.base_experiment import BaseExperiment
from matipo.experiment.plots import SignalPlot, SpectrumPlot

# Import panel library and enable panel extension.
# This is required when using the experiment library for the GUI elements to display in the jupyter notebook.
import panel as pn
pn.extension(inline=True)

# Simple Pulse Collect Experiment that uses the default parameters from the pulse sequence program file

class SignalAndSpectrum(BaseExperiment):
    def setup(self):
        
        # Set workspace for data to be saved to
        self.workspace = "example"
        
        # load the pulse sequence from global pulse sequence directory
        self.seq = Sequence(SEQUENCE_DIR+'FID.py')
        
        # Create the signal and spectrum plots
        self.plots = {
            'signal': SignalPlot(),
            'spectrum': SpectrumPlot()
        }

    # Define how plots are updated with data, this function is called automatically once the sequence has finished
    async def update_plots(self, *args, **kwargs):
        await self.seq.fetch_data()
        self.plots['signal'].update(seqdata=self.seq.data, t_dw=self.seq.par.t_dw)
        self.plots['spectrum'].update(seqdata=self.seq.data, t_dw=self.seq.par.t_dw, t0=0)
            
# Create experiment object
exp1 = SignalAndSpectrum(state_id='tut2exp1')

# Return the GUI for display
exp1()
```

## 1D Imaging

The following code is for a very simple implemtation of a 1D Image experiment.

In this experiment, the Spin Echo pulse sequence (`SE.py`) is loaded from *ilumr's* global sequence directory.

The experiment uses most of the default pulse sequence parameters in the `SE.py` pulse sequence file, with the exception of the read gradient parameter (`g_read`). In the default SE pulse sequence, the read gradient is set to (0,0,0) but, for this experiment, a gradient of 0.1 T/m is set along the Z axis. `g_read` is not set in T/m but is, rather, a unitless value representing the proportion of the maximum gradient that must be used. The value (`G_CAL`) needed to convert 0.1 T/m to this proportional value is calculated using the gyromagnetic ratio of the nuclei (`GAMMA_BAR`) and the value loaded from `gradient_calibration.yaml`. This `G_CAL` value is the gradient amplitude in T/m when `g_read` is set to 1 (the maximum value) along an axis.

The data is displayed using the *matipo* experiment library's `Image1DPlot`. This plot type is similar to the spectrum plot but automatically scales the frequency encode axis to units of meters based on the read gradient strength. To do this, the plot's update function must be passed the magnitude of the read gradient.

```python
import os
import yaml
import numpy as np
from matipo import Sequence, SEQUENCE_DIR, GLOBALS_DIR
from matipo.experiment.base_experiment import BaseExperiment
from matipo.experiment.plots import Image1DPlot

# Import panel library and enable panel extension.
# This is required when using the experiment library for the GUI elements to display in the jupyter notebook.
import panel as pn
pn.extension(inline=True)

GAMMA_BAR = 42.58e6 # 1H gyromagnetic ratio in Hz/T

class Image1D(BaseExperiment):
    
    def setup(self):
        
        # Set workspace for data to be saved to
        self.workspace = 'example'
        
        # load the pulse sequence from global pulse sequence directory
        self.seq = Sequence(SEQUENCE_DIR+'SE.py')
        
        # Create the 1D image plot
        self.plots = {
            'image': Image1DPlot()
        }
        
        # Load global gradient calibration value and convert to T/m
        # G_CAL is the T/m at full amplitude
        with open(os.path.join(GLOBALS_DIR, 'gradient_calibration.yaml'), 'r') as f:
            data = yaml.load(f, Loader=yaml.SafeLoader)
            self.G_CAL = 1/(GAMMA_BAR*data['gradient_calibration']) 
            self.log.debug(f'G_CAL: {self.G_CAL}')

    def update_par(self):
        self.seq.setpar(
            # Set 0.1 T/m gradient on Z axis.
            # G_CAL is used convert the gradient from physical units to a
            # proportional value used by the pulse sequence
            g_read=(0,0,0.1/self.G_CAL),
        )
    
    # Define how plots are updated with data, this function is called automatically once the sequence has finished
    async def update_plots(self, *args, **kwargs):
        await self.seq.fetch_data()
        self.plots['image'].update(
            seqdata=self.seq.data,
            t_dw=self.seq.par.t_dw,
            # g_read_mag needs to be in T/m, so convert the proportional
            # g_read vector to a magnitude in physical units
            g_read_mag=self.G_CAL*np.linalg.norm(self.seq.par.g_read),
        )
            
exp2 = Image1D(state_id='tut2exp2')
exp2()
```

## 2D Imaging with k-Space and Image Plots

The code below provides an example for setting up a 2D RARE imaging experiments with k-space and image plots. 

In this experiment, the Rapid Aquisition with Relaxation Enhancement pulse sequence (`RARE.py`) is loaded from *ilumr's* global sequence directory.

The pulse sequence parameters are overidden in the `setup` method. 

Both the k-space and image plot use `ImagePlot` from the matipo experiment library. 

`ImagePlot` plots 2D data as an image but does not handle axis titles or data processing. 

Therefore, we need to:
- Set the k-space plot title
- Set all axis labels
- Calculate the field of view (FOV) in each axis
- Reshape the raw signal data into a 2D array
- Define the post processing decimation factor and apply it to the data
- Convert the k-space data into image data using a fast fourier transform
- Update each plot with the relevant data and scaled FOV

Each of these steps are commented in the code below. One step that is worth expanding on is the post processing decimation.

Because *ilumr's* hardware filter has a non-flat frequency response, extra samples are aquired and a flat decimation filter is applied to select the center of the aquisition - resulting in flat frequency profile.

The decimation factor `self.DEC` is set in the experiment and then used to calculate the number of samples (`n_samples`) and dwell time (`t_dwell`) needed for oversampling. After the raw signal has been shaped into a 2D array, the flat filter is applied using the `decimate()` function. This data is then used to update the k-space plot.

```python
import os
import yaml
import numpy as np
from matipo import Sequence, SEQUENCE_DIR, GLOBALS_DIR
from matipo.experiment.base_experiment import BaseExperiment
from matipo.experiment.plots import ImagePlot
from matipo.util.fft import fft_reconstruction
from matipo.util.decimation import decimate

# Import panel library and enable panel extension.
# This is required when using the experiment library for the GUI elements to display in the jupyter notebook.
import panel as pn
pn.extension(inline=True)

GAMMA_BAR = 42.58e6 # 1H gyromagnetic ratio in Hz/T

class Image2D(BaseExperiment):
    def setup(self):
        
        # Set workspace for data to be saved to
        self.workspace = 'example'
        
        # load the pulse sequence from global pulse sequence directory
        self.seq = Sequence(SEQUENCE_DIR+'RARE.py')
        self.enable_partialplot = True
        
        # Create the 2D k-space and image plots
        self.plots = {
            'kspace': ImagePlot(
                figure_opts=dict(
                    title='k-space',
                    x_axis_label=r'Freq. Encode Axis (m^-1)',
                    y_axis_label=r'Phase. Encode Axis (m^-1)'
                ),
                image_opts=dict(palette='Viridis256')
            ),
            'image': ImagePlot(
                figure_opts=dict(
                    x_axis_label='Freq. Encode Axis (mm)',
                    y_axis_label='Phase. Encode Axis (mm)'
                )
            )
        }

        # Load global gradient calibration value and convert to T/m
        # G_CAL is the T/m at full amplitude
        with open(os.path.join(GLOBALS_DIR, 'gradient_calibration.yaml'), 'r') as f:
            data = yaml.load(f, Loader=yaml.SafeLoader)
            self.G_CAL = 1/(GAMMA_BAR*data['gradient_calibration']) # convert to T/m
            self.log.debug(f'G_CAL: {self.G_CAL}')

        self.DEC = 4 # postprocess decimation factor

        # Set nice default pars
        self.seq.setpar(
            n_scans=4,
            t_end=0.5,
            n_phase_1 = 1,
            n_phase_2 = 32,
            g_phase_2 = (0, 0.5, 0),
            g_read = (0.5, 0, 0),
            
            # Oversample by a factor of DEC so that we can use the flat decimation filter
            t_dw=20e-6/self.DEC,
            n_samples=32*self.DEC,
            
            t_phase=320e-6
        )

        # Set calculated pars
        self.seq.setpar(
            n_ETL=self.seq.par.n_phase_2,
            t_read=self.seq.par.t_dw*self.seq.par.n_samples
        )

    # Calculate FOV in read direction
    def fov_read(self, gamma_bar=GAMMA_BAR):
        """ Returns read encode direction FOV """
        dt = self.seq.par.t_dw*self.DEC
        g = self.G_CAL*np.linalg.norm(self.seq.par.g_read)
        return 1/(g*gamma_bar*dt)

    # Calculate FOV in phase direction
    def fov_phase(self, gamma_bar=GAMMA_BAR):
        """ Returns phase encode direction FOV """
        t = self.seq.par.t_phase
        dg = self.G_CAL*np.linalg.norm(self.seq.par.g_phase_2)/(self.seq.par.n_phase_2//2)
        return 1/(dg*gamma_bar*t)

    # Define how plots are updated with data, this function is called automatically once the sequence has finished
    async def update_plots(self, final):
        await self.seq.fetch_data()
        
        # Reshape the one dimensional array of all data into a two dimensional array of individual acquisitions
        raw_data_2d = self.seq.data.reshape(-1, self.seq.par.n_samples)
        
        # Apply a flat filter decimation to each aquisition,
        # this compensates for the non-flat frequency response of ilumr's hardware filter
        kdata = decimate(raw_data_2d, self.DEC, axis=1)
        
        # Use fast fourier transform to convert k-space data into image data
        # A small amount of gaussian blur is applied to reduce the Gibbs artifact
        imdata = fft_reconstruction(kdata, gaussian_blur=0.5)
        
        fov_read = self.fov_read()
        fov_phase = self.fov_phase()
        
        # Use processed data to update k-space and image plots
        self.plots['kspace'].update(np.abs(kdata), kdata.shape[1]/fov_read, kdata.shape[0]/fov_phase)
        self.plots['image'].update(np.abs(imdata), 1e3*fov_read, 1e3*fov_phase) # Convert FOV to mm to match axis labels
        
        
exp3 = Image2D(state_id='tut2exp3')
exp3()
```
