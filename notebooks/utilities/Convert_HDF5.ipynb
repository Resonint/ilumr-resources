# Converting HDF5 Data Files to Other Formats

Files saved by experiments using the `BaseExperiment` class are HDF5 files with the acquisition data and pulse sequence parameters both included as datasets.
These can be loaded with the python "h5py" module as shown below, and saved as other formats if necessary.
It is also possible to override the `BaseExperment.save()` method when creating/modifying an experiment in order to save in these other formats directly ([docs](https://resonint-kahu.readthedocs-hosted.com/en/latest/matipo/api.html#matipo.experiment.BaseExperiment.save)).

```python
from matipo import DATA_DIR
FILEPATH = DATA_DIR+'default/CPMG/CPMG_20251119_0_doped_water_80us.hdf5'

import h5py
import yaml

with h5py.File(FILEPATH, 'r') as f:
    data = f['dataset']['data'][()]
    par = yaml.full_load(f['dataset']['par_yaml'][()])
data, par
```

```python
# save data to npy format, to same directory and file name as was loaded
import numpy as np
np.save(FILEPATH.replace('.hdf5','.npy'), data)
```

```python
# save data and parameters to npz format, to same directory and file name as was loaded
import numpy as np
np.savez(FILEPATH.replace('.hdf5','.npz'), data=data, par=par)
```

```python
# save data and parameters to matlab format, to same directory and file name as was loaded
import scipy as sp
sp.io.savemat(FILEPATH.replace('.hdf5','.mat'), dict(data=data, par=par))
```
